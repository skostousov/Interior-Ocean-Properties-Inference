```python
from utils.config import DATA_DIR, SAVED_MODELS_DIR, RAW_CONFIG
from pathlib import Path
from utils.dataset2 import GLORYSDS2
import torch
import numpy as np
from utils.handy_plotting_func import plot_var
```

```python
data_norm = GLORYSDS2(dataset_dir = Path(DATA_DIR/RAW_CONFIG["datafile"]), normalize=True)
data = GLORYSDS2(dataset_dir = Path(DATA_DIR/RAW_CONFIG["datafile"]))
```

```python
from matplotlib import pyplot as plt
num_samples = 50
fig, axes = plt.subplots(num_samples, 3, figsize=(15, 5*num_samples))
for i in range(num_samples):
    _, original_label = data[i]
    _, label_norm = data_norm[i]
    max = 8*1e-5
    plot_var(axes[i, 0], fig, original_label[0], "MLD", max=None)
    plot_var(axes[i, 1], fig, label_norm[0], "Normalized MLD", max=None)
    plot_var(axes[i, 2], fig, np.squeeze(data_norm.unnormalize_annot(np.expand_dims(label_norm, 0)), (0, 1)), "Unnormalized MLD", max=None)
plt.show()
```
