```python
from utils.datasetmonths import DatasetOverMonths

ds = DatasetOverMonths()
```

```python
from utils.splitter import train_val_test_split_months_strat
train_idx, val_idx, test_idx = train_val_test_split_months_strat(ds)
```

```python
from torch.utils.data import Subset
from utils.datasetmonths import TestSubsetRegression

train_ds = Subset(ds, train_idx)
val_ds = Subset(ds, val_idx)
test_ds = TestSubsetRegression(ds, test_idx)
print(f"Train dataset size: {len(train_ds)}")
print(f"Validation dataset size: {len(val_ds)}")
print(f"Test dataset size: {len(test_ds)}")
```

```python
months_train_idx, months_val_idx, months_test_idx = [ds.groups_by_month[i] for i in train_idx], [ds.groups_by_month[i] for i in val_idx], [ds.groups_by_month[i] for i in test_idx]
months_dated_train_idx, months_dated_val_idx, months_dated_test_idx = [ds.groups[i] for i in train_idx], [ds.groups[i] for i in val_idx], [ds.groups[i] for i in test_idx]
```
