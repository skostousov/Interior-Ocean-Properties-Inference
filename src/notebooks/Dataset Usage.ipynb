```python
import numpy as np

from utils.dataset import GLORYSDS
from utils.config import DATA_DIR
```

```python
filename = 'cmems_mod_glo_phy_my_0.083deg_P1D-m_multi-vars_156.00W-121.42W_41.83N-63.08N_0.49m_2021-06-25-2021-06-30.nc'

data_dir =DATA_DIR/filename
```

```python
dataset = GLORYSDS(data_dir, grid_size=100)
mlds = dataset[2][1]
mlds_squeezed = np.squeeze(dataset[2][1], axis=1)
```

```python
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display
import numpy as np


slider = widgets.IntSlider(value=0, min=0, max=mlds_squeezed.shape[0]-1, step=1, description='Day')

def plot_var(axis, fig, day, var, var_name):
    vmin = 0
    vmax = np.max(var)
    axis.set_xlabel("Longitude")
    axis.set_ylabel("Latitude")
    im = axis.imshow(var[day], origin='lower', aspect='auto', cmap='viridis', vmin=vmin, vmax=vmax)
    axis.set_xticks([])
    axis.set_yticks([])
    fig.colorbar(im, ax=axis, label=var_name)

def plot_day(day):
    fig, axs = plt.subplots(figsize=(10, 10))
    plot_var(axs, fig, day, mlds_squeezed, "Depth of Mixed Ocean Layer (m)")

widgets.interact(plot_day, day=slider)
```
